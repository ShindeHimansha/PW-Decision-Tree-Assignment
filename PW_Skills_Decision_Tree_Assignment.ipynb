
#1. What is a Decision Tree, and how does it work
-	A Decision Tree is a supervised machine learning algorithm used for both classification and regression tasks.
-	It works by recursively splitting the dataset into subsets based on feature values that result in the most information gain or least impurity.
-	Each internal node of the tree represents a test on a feature, each branch represents the outcome of the test, and each leaf node represents a final output label or value.
-	The goal is to create a model that predicts the target variable by learning simple decision rules from data features.

#2. What are impurity measures in Decision Trees
-	Impurity measures are mathematical methods used to determine the quality of a split in a decision tree.
-	These measures quantify the level of disorder or uncertainty in the target variable distribution of a node.
-	Common impurity measures include Gini Impurity and Entropy, which are used to evaluate how mixed the classes are in a node.
-	A lower impurity indicates a better split and helps the tree learn more accurate decision boundaries.

#3. What is the mathematical formula for Gini Impurity
-	Gini Impurity measures the probability of incorrectly classifying a randomly chosen element from a dataset.
-	The formula is:
Gini(p)=1−∑i=1npi2Gini(p) = 1 - \sum_{i=1}^{n} p_i^2
-	Here, pip_i is the probability of an element being classified into a particular class ii, and the sum is over all classes.
-	A Gini value of 0 indicates perfect purity, meaning all elements belong to one class.

#4. What is the mathematical formula for Entropy
-	Entropy is a measure from information theory that quantifies the amount of uncertainty or impurity in a dataset.
-	The formula is:
Entropy(p)=−∑i=1npilog⁡2(pi)Entropy(p) = -\sum_{i=1}^{n} p_i \log_2(p_i)
-	Where pip_i is the probability of class ii in the node, and the sum is taken over all classes.
-	Entropy is 0 when the node is pure (only one class is present) and reaches its maximum when classes are equally distributed.

#5. What is Information Gain, and how is it used in Decision Trees
-	Information Gain is the reduction in impurity or entropy achieved by partitioning the dataset based on a given feature.
-	It is calculated as the difference between the impurity of the parent node and the weighted impurity of child nodes.
-	Decision Trees use Information Gain to select the best feature to split the data at each node, aiming to maximize it.
-	Higher Information Gain indicates a more useful feature for class separation.

#6. What is the difference between Gini Impurity and Entropy
-	Both Gini Impurity and Entropy are impurity measures, but they use different mathematical formulations.
-	Gini Impurity is computationally simpler and measures the probability of incorrect classification, while Entropy measures the level of information or surprise.
-	Although both often lead to similar tree structures, Gini tends to create more balanced splits and is faster to compute.
-	Entropy tends to produce more pure nodes at the cost of more computation due to the logarithmic component.

#7. What is the mathematical explanation behind Decision Trees
-	Decision Trees use recursive binary splitting to partition the input space into regions with mostly homogeneous target values.
-	At each node, the algorithm selects a feature and threshold that yields the highest Information Gain or lowest Gini Impurity.
-	The process continues until a stopping criterion is met such as maximum depth, minimum samples per leaf, or zero impurity.
-	The final prediction is made by traversing the tree based on feature values until a leaf node is reached.

#8. What is Pre-Pruning in Decision Trees
-	Pre-Pruning is a technique used to prevent overfitting by stopping the tree growth early during training.
-	It involves setting constraints such as maximum tree depth, minimum samples per split, or minimum impurity decrease.
-	If a node fails to meet these criteria, it becomes a leaf node without further splitting.
-	This approach simplifies the model and often improves generalization on unseen data.

#9. What is Post-Pruning in Decision Trees
-	Post-Pruning involves building a full decision tree and then trimming it after training to remove nodes that do not improve performance.
-	The goal is to reduce model complexity and avoid overfitting while retaining important decision rules.
-	This is done by evaluating the tree’s accuracy on a validation set and pruning subtrees that do not contribute to improved accuracy.
-	Cost Complexity Pruning is a common post-pruning technique used in scikit-learn.

#10. What is the difference between Pre-Pruning and Post-Pruning
-	Pre-Pruning stops the tree growth early by applying constraints during the training phase, while Post-Pruning allows full growth before trimming.
-	Pre-Pruning is computationally efficient as it avoids unnecessary growth, but may miss some beneficial splits.
-	Post-Pruning provides more control and often leads to better model performance by evaluating actual results on validation data.
-	Pre-Pruning is preventive; Post-Pruning is corrective.

#11. What is a Decision Tree Regressor
-	A Decision Tree Regressor is a variant of a Decision Tree used for predicting continuous numerical values rather than categories.
-	Instead of minimizing classification impurity, it minimizes the mean squared error (MSE) within leaf nodes.
-	It works by recursively splitting the data into subsets that are as homogeneous as possible in terms of target variable.
-	The final prediction is the average of values in the leaf node reached by an input sample.

#12. What are the advantages and disadvantages of Decision Trees
-	Advantages:
    -	Easy to understand and interpret.
    -	Can handle both numerical and categorical data.
    -	Requires little data preprocessing and can model non-linear relationships.
-	Disadvantages:
    -	Prone to overfitting especially with deep trees.
    -	Unstable to small changes in data.
    -	Greedy algorithm may not yield the globally optimal tree.

#13. How does a Decision Tree handle missing values
-	Decision Trees can handle missing values by ignoring them during the split or using surrogate splits.
-	Surrogate splits use alternative features that closely mimic the splitting behavior of the primary feature.
-	Some implementations impute missing values based on statistics such as the most frequent value or mean.
-	This ensures the model remains robust even with incomplete data.

#14. How does a Decision Tree handle categorical features
-	Decision Trees can handle categorical features by evaluating splits based on the different categories directly.
-	For each categorical feature, the algorithm checks each possible grouping of categories to find the best split.
-	In some libraries like scikit-learn, categorical features need to be encoded numerically using methods like one-hot encoding.
-	Advanced implementations can directly split on category labels without numeric encoding.

#15. What are some real-world applications of Decision Trees?
-	Medical diagnosis for determining diseases based on symptoms.
-	Customer segmentation in marketing for targeted advertising.
-	Credit scoring and fraud detection in financial institutions.
-	Recommendation systems for personalizing content.
-	Manufacturing and supply chain optimization decisions.



```python
#1. Write a Python program to train a Decision Tree Classifier on the Iris dataset and print the model accuracy

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

iris = load_iris()
X, y = iris.data, iris.target

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

clf = DecisionTreeClassifier()
clf.fit(X_train, y_train)

y_pred = clf.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
```

```python
#2. Write a Python program to train a Decision Tree Classifier using Gini Impurity as the criterion and print the feature importances

from sklearn.tree import DecisionTreeClassifier

clf_gini = DecisionTreeClassifier(criterion='gini')
clf_gini.fit(X_train, y_train)

print("Feature importances (Gini):", clf_gini.feature_importances_)
```

```python
#3. Write a Python program to train a Decision Tree Classifier using Entropy as the splitting criterion and print the model accuracy

clf_entropy = DecisionTreeClassifier(criterion='entropy')
clf_entropy.fit(X_train, y_train)

y_pred_entropy = clf_entropy.predict(X_test)
print("Accuracy (Entropy):", accuracy_score(y_test, y_pred_entropy))
```

```python
#4. Write a Python program to train a Decision Tree Regressor on a housing dataset and evaluate using Mean Squared Error (MSE)

from sklearn.datasets import fetch_california_housing
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error

housing = fetch_california_housing()
X_h, y_h = housing.data, housing.target

X_train_h, X_test_h, y_train_h, y_test_h = train_test_split(X_h, y_h, test_size=0.3, random_state=42)

reg = DecisionTreeRegressor()
reg.fit(X_train_h, y_train_h)

y_pred_h = reg.predict(X_test_h)
print("Mean Squared Error:", mean_squared_error(y_test_h, y_pred_h))
```

```python
#12. Write a Python program to train a Decision Tree Classifier, apply Cost Complexity Pruning (CCP), and visualize its effect on accuracy

path = clf.cost_complexity_pruning_path(X_train, y_train)
ccp_alphas = path.ccp_alphas

acc_scores = []
for ccp_alpha in ccp_alphas:
    clf_pruned = DecisionTreeClassifier(ccp_alpha=ccp_alpha)
    clf_pruned.fit(X_train, y_train)
    acc_scores.append(accuracy_score(y_test, clf_pruned.predict(X_test)))

import matplotlib.pyplot as plt

plt.plot(ccp_alphas, acc_scores, marker='o')
plt.xlabel("ccp_alpha")
plt.ylabel("Accuracy")
plt.title("Effect of CCP on Accuracy")
plt.grid(True)
plt.show()
```
